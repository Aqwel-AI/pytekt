# Machine-learning maths

Activation functions, prediction losses, and vector similarity metrics.

```python
from aion.maths.machine_learning import (
    cosine_similarity, cross_entropy_loss, euclidean_distance,
    relu, sigmoid, softmax,
)
```

## Activations

```python
scores = [-2, 0, 2]
print('Sigmoid:', sigmoid(scores))
print('ReLU:', relu(scores))
print('Softmax:', softmax(scores))
```

## Losses and distances

```python
truth = [1, 0, 1]
predictions = [0.9, 0.1, 0.8]
print('Cross-entropy:', cross_entropy_loss(truth, predictions))
print('Euclidean distance:', euclidean_distance([1, 2], [4, 6]))
print('Cosine similarity:', cosine_similarity([1, 1], [1, 1]))
```

## More activations and losses

```python
from aion.maths.machine_learning import leaky_relu, mae_loss, mse_loss, tanh_activation

print('Tanh:', tanh_activation([-1, 0, 1]))
print('Leaky ReLU:', leaky_relu([-2, 0, 2]))
print('MSE:', mse_loss([1, 2, 3], [1.1, 2.1, 2.9]))
print('MAE:', mae_loss([1, 2, 3], [1.1, 2.1, 2.9]))
```
