# Trigonometry and signal processing

Work with angles, discrete Fourier transforms, and convolution.

```python
import math
from aion.maths.trigonometry import cos, degrees, radians, sin
from aion.maths.signal_processing import convolution, fft, ifft
```

## Angles

```python
angle = radians(45)
print('sin(45°):', sin(angle))
print('cos(45°):', cos(angle))
print('Back to degrees:', degrees(angle))
assert math.isclose(sin(angle), math.sqrt(0.5))
```

## Fourier transform and convolution

```python
signal = [1, 0, 1, 0]
coefficients = fft(signal)
print('FFT:', coefficients)
print('Inverse FFT:', ifft(coefficients))
print('Convolution:', convolution([1, 2, 3], [1, 1]))
```

## More trigonometry

```python
from aion.maths.trigonometry import acos, asin, atan

print('asin(1):', asin(1))
print('acos(1):', acos(1))
print('atan(1):', atan(1))
```
