# Linear algebra

Vectors, matrices, matrix products, inverses, and decompositions.

```python
from aion.maths.linear_algebra import (
    determinant, dot_product, matrix_inverse, matrix_multiply,
    normalize_vector, vector_magnitude,
)
```

## Vectors

```python
u = [3, 4]
v = [1, 2]
print('Dot product:', dot_product(u, v))
print('Magnitude:', vector_magnitude(u))
print('Unit vector:', normalize_vector(u))
```

## Matrices

```python
a = [[1, 2], [3, 4]]
b = [[5, 6], [7, 8]]
print('Product:', matrix_multiply(a, b))
print('Determinant:', determinant(a))
print('Inverse:', matrix_inverse(a))
```

The matrix functions return ordinary Python lists, which makes them convenient for small educational and research examples.

## More matrix operations

```python
from aion.maths.linear_algebra import cross_product, eigenvalues, matrix_rank, transpose

print('Transpose:', transpose(a))
print('Eigenvalues:', eigenvalues(a))
print('Rank:', matrix_rank(a))
print('Cross product:', cross_product([1, 0, 0], [0, 1, 0]))
```
