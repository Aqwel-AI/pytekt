# Arithmetic and number theory

Use focused category imports for everyday calculations and integer mathematics.

```python
from aion.maths.arithmetic import addition, division, factorial, log, power
from aion.maths.number_theory import fibonacci, is_prime, prime_factors
```

## Basic arithmetic

```python
print('2 + 3 =', addition(2, 3))
print('12 / 4 =', division(12, 4))
print('2**8 =', power(2, 8))
print('10! =', factorial(10))
print('log(e) =', log(2.718281828459045))
```

## Prime factors and sequences

```python
number = 84
print(number, 'is prime:', is_prime(number))
print('Prime factors:', prime_factors(number))
print('Fibonacci(10):', fibonacci(10))
```

The original flat API remains available as `aion.maths.addition(...)`; category imports make the source of each function explicit.

## More arithmetic

Arithmetic functions also support vector-like inputs and common integer helpers.

```python
from aion.maths.arithmetic import gcd, lcm, sqrt

print('GCD(48, 18):', gcd(48, 18))
print('LCM(4, 6):', lcm(4, 6))
print('Vector addition:', addition([1, 2, 3], 10))
print('Square root:', sqrt(81))
```
