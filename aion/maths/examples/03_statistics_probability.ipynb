# Statistics and probability

Summarize data, standardize values, and evaluate common distributions.

```python
from aion.maths.statistics import correlation, mean, std_dev, variance, z_score
from aion.maths.probability import binomial_pmf, normal_cdf, normal_pdf, poisson_pmf
```

## Descriptive statistics

```python
values = [2, 4, 4, 6, 8]
print('Mean:', mean(values))
print('Variance:', variance(values))
print('Standard deviation:', std_dev(values))
print('Z-scores:', z_score(values))
print('Correlation:', correlation(values, [1, 2, 3, 4, 5]))
```

## Probability mass and density functions

```python
print('Normal PDF at 0:', normal_pdf(0))
print('Normal CDF at 0:', normal_cdf(0))
print('Binomial P(X=2):', binomial_pmf(2, n=5, p=0.5))
print('Poisson P(X=2):', poisson_pmf(2, lam=3.0))
```

## Scaling and regression

```python
from aion.maths.statistics import linear_regression, median, min_max_scale

print('Median:', median(values))
print('Min-max scale:', min_max_scale(values))
print('Regression slope/intercept:', linear_regression([1, 2, 3], [2, 4, 6]))
```
