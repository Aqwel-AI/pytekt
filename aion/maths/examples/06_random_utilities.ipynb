# Random sampling and utilities

Make experiments reproducible and use small numerical helpers.

```python
from aion.maths.random import sample_normal, sample_uniform, set_seed, train_test_split
from aion.maths.utilities import clamp, lerp
```

## Reproducible samples

```python
set_seed(42)
print('Uniform:', sample_uniform(0, 1, size=3))
print('Normal:', sample_normal(size=3))
train, test = train_test_split(list(range(10)), ratio=0.8)
print('Train:', train)
print('Test:', test)
```

## Utility functions

```python
print('Clamp -2 to [0, 10]:', clamp(-2, 0, 10))
print('Interpolate from 10 to 20 at t=0.25:', lerp(10, 20, 0.25))
```

## Choices and shuffling

```python
from aion.maths.random import random_choice, shuffle_list

set_seed(7)
print('Weighted choice index:', random_choice([0.1, 0.7, 0.2]))
print('Shuffled values:', shuffle_list([1, 2, 3, 4, 5]))
```
