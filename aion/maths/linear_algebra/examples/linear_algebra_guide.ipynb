# Aion Linear Algebra Guide
This notebook demonstrates the usage of the reorganized `aion.maths.linear_algebra` module, showing both vector and matrix operations.

## 1. Vector Operations
We can perform common vector math operations like addition, subtraction, dot product, cross product, magnitude, and normalization.

```python
from aion.maths.linear_algebra import (
    vector_add,
    vector_subtract,
    vector_scale,
    dot_product,
    cross_product,
    vector_magnitude,
    normalize_vector,
    angle_between_vectors
)

# Define two 3D vectors
u = [1.0, 2.0, 3.0]
v = [4.0, 5.0, 6.0]

print("u:", u)
print("v:", v)
print("u + v:", vector_add(u, v))
print("u - v:", vector_subtract(u, v))
print("2 * u:", vector_scale(u, 2.0))
print("Dot Product:", dot_product(u, v))
print("Cross Product:", cross_product(u, v))
print("Magnitude of u:", vector_magnitude(u))
print("Normalized u:", normalize_vector(u))
print("Angle (rad) between u and v:", angle_between_vectors(u, v))
```

## 2. Matrix Operations
The library offers extensive matrix support, including generators, transpose, multiplication, solvers, and decompositions.

```python
from aion.maths.linear_algebra import (
    zeros,
    identity,
    transpose,
    matrix_multiply,
    determinant,
    matrix_inverse,
    solve_linear_system
)

# Create matrices
A = [[2.0, 1.0], [1.0, 3.0]]
B = [[1.0, 0.0], [0.0, 1.0]]

print("A:", A)
print("A transpose:", transpose(A))
print("A * B:", matrix_multiply(A, B))
print("Det(A):", determinant(A))
print("A inverse:", matrix_inverse(A))

# Solve Ax = b
b = [8.0, 9.0]
x = solve_linear_system(A, b)
print("Solution x to Ax = b:", x)
```
