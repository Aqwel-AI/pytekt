# Unified Catalog Browser & Introspection

The algorithm catalog manages metadata for over 570+ functions across 7 domain categories.


## 1. Global Registry Overview


```python
from pytekt.algorithms import count_algorithms, categories, list_algorithms

print("Total catalogued algorithms:", count_algorithms())
print("Registered categories:", categories())
```

## 2. Category Search & Filtering


```python
from pytekt.algorithms import categories, list_algorithms

# Inspect algorithm counts per category
for cat in categories():
    entries = list_algorithms(cat)
    print(f"{cat:<22} : {len(entries):>3} algorithms")
```

## 3. Search by Name or Keyword


```python
from pytekt.algorithms import list_algorithms

keyword = "tree"
matches = [e for e in list_algorithms() if keyword in e.name.lower() or keyword in e.summary.lower()]
print(f"Found {len(matches)} algorithms matching '{keyword}':\n")

for e in matches[:8]:
    print(f"[{e.category}] {e.name}: {e.summary}")
```
