# Catalog browser

Search the full algorithms registry.

```python
from pytekt.algorithms import categories, count_algorithms, list_algorithms

print("Total algorithms:", count_algorithms())
print("Categories:", categories())

needle = "palindrome"
matches = [e for e in list_algorithms() if needle in e.name or needle in e.summary.lower()]
for e in matches[:10]:
    print(f"{e.category}.{e.name}: {e.summary}")
```
