# Data Structures: Trees, Heaps, Stacks & Union-Find

Demonstrates advanced and foundational structures from `pytekt.algorithms.data_structures`.


## 1. Binary Search Trees & Tree Algorithms


```python
from pytekt.algorithms.data_structures.trees import (
    TreeNode,
    build_bst_from_sorted,
    bst_insert,
    bst_search,
    inorder_traversal,
    tree_height,
)

# Build balanced BST
values = [10, 20, 30, 40, 50, 60, 70]
root = build_bst_from_sorted(values)

print("In-order traversal:", inorder_traversal(root))
print("Tree height:       ", tree_height(root))
print("Search 40 in BST:  ", bst_search(root, 40) is not None)
```

## 2. Heaps & Priority Queues


```python
from pytekt.algorithms.data_structures.heaps import (
    build_max_heap,
    heapify_list,
    heappush_wrapper,
    heappop_wrapper,
)

h = [15, 10, 20, 8, 12]
heapify_list(h)
print("Min-heapified list:", h)
print("Pop min element:   ", heappop_wrapper(h))
```

## 3. Disjoint Set Union (Union-Find) & Cycle Detection


```python
from pytekt.algorithms.data_structures.union_find import (
    make_disjoint_set,
    union_sets,
    connected,
    count_components,
    detect_cycle_undirected,
)

n = 5
parent = make_disjoint_set(n)
rank = [0] * n

union_sets(parent, rank, 0, 1)
union_sets(parent, rank, 1, 2)
union_sets(parent, rank, 3, 4)

print("Connected (0, 2):   ", connected(parent, 0, 2))
print("Connected (0, 3):   ", connected(parent, 0, 3))
print("Number of components:", count_components(n, [(0, 1), (1, 2), (3, 4)]))
print("Cycle detected:     ", detect_cycle_undirected(3, [(0, 1), (1, 2), (2, 0)]))
```
