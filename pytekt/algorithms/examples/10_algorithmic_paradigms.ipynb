# Algorithmic Paradigms: Backtracking, Two Pointers & Bit Manipulation

Explore advanced algorithm design patterns from `pytekt.algorithms.paradigms`.


## 1. Backtracking & Combinatorics


```python
from pytekt.algorithms.paradigms.backtracking import (
    n_queens_solutions,
    subsets,
    permutations,
    combination_sum,
)

print("Subsets of [1, 2]:")
print(subsets([1, 2]))

print("\nPermutations of [1, 2, 3] (first 3):")
print(permutations([1, 2, 3])[:3])

print("\nN-Queens solutions for N=4:")
for sol in n_queens_solutions(4):
    print(sol)
```

## 2. Two Pointers & Sliding Window


```python
from pytekt.algorithms.paradigms.two_pointers import (
    container_with_most_water,
    two_sum_sorted,
    trap_rain_water,
)
from pytekt.algorithms.paradigms.sliding_window import (
    max_sum_subarray_size_k,
    longest_substring_without_repeating,
)

# Container with most water
heights = [1, 8, 6, 2, 5, 4, 8, 3, 7]
print("Max water container area:", container_with_most_water(heights))

# Trapping rain water
elevations = [0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1]
print("Trapped rain water:     ", trap_rain_water(elevations))

# Sliding Window Maximum Sum of window size 3
print("Max sum subarray k=3:   ", max_sum_subarray_size_k([2, 1, 5, 1, 3, 2], 3))
print("Longest substring w/o repeating ('abcabcbb'):", longest_substring_without_repeating("abcabcbb"))
```

## 3. Bit Manipulation Tricks


```python
from pytekt.algorithms.paradigms.bit_manipulation import (
    popcount,
    is_power_of_two,
    single_number,
    hamming_distance,
)

print("Set bits in 29 (0b11101):", popcount(29))
print("Is 64 a power of two?    ", is_power_of_two(64))
print("Is 60 a power of two?    ", is_power_of_two(60))
print("Single non-duplicate in [4, 1, 2, 1, 2]:", single_number([4, 1, 2, 1, 2]))
print("Hamming distance (1, 4): ", hamming_distance(1, 4))
```
