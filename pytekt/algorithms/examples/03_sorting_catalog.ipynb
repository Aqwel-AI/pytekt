# Sorting catalog

Browse and run registered sorting algorithms via `get_algorithm`.

```python
from pytekt.algorithms import get_algorithm, list_algorithms

for entry in list_algorithms("sorting")[:8]:
    print(entry.name, "—", entry.summary)

data = [3, 1, 4, 1, 5, 9, 2, 6]
merge_sort = get_algorithm("merge_sort")
print("merge_sort:", merge_sort(data))
```
