# Sorting Algorithms & Catalog Exploration

This notebook demonstrates the sorting algorithms available in `pytekt.algorithms.sorting` as well as dynamic lookup via the unified catalog.


## 1. Direct Subpackage Sorting Algorithms

Import classic comparison and distribution sorting routines directly from `pytekt.algorithms.sorting`.


```python
from pytekt.algorithms.sorting import (
    bubble_sort,
    selection_sort,
    insertion_sort,
    merge_sort,
    quick_sort,
    heap_sort,
    counting_sort,
    radix_sort,
    bucket_sort,
    shell_sort,
    tim_sort,
)

data = [34, 12, 89, 5, 23, 77, 1, 90, 45]

print("Original array:  ", data)
print("Quick sort:      ", quick_sort(data))
print("Merge sort:      ", merge_sort(data))
print("Heap sort:       ", heap_sort(data))
print("Radix sort:      ", radix_sort(data))
print("Shell sort:      ", shell_sort(data))
print("Tim sort:        ", tim_sort(data))
```

## 2. Dynamic Catalog Discovery for Sorting

Explore all registered sorting routines, metadata, and doc summaries dynamically.


```python
from pytekt.algorithms import list_algorithms, get_algorithm

sorting_algos = list_algorithms("sorting")
print(f"Total registered sorting algorithms: {len(sorting_algos)}\n")

for entry in sorting_algos[:10]:
    print(f"• {entry.name:<25} [{entry.module}] — {entry.summary}")

# Execute dynamically
fn = get_algorithm("counting_sort")
print("\nCounting sort result:", fn([4, 2, 2, 8, 3, 3, 1]))
```
