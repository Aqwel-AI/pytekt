# Strings Processing & Data Compression

Explore string patterns, analysis, and data compression from `pytekt.algorithms.strings`.


## 1. String Utilities & Pattern Matching


```python
from pytekt.algorithms.strings.strings import (
    longest_palindrome_substring,
    is_anagram,
    count_palindromes,
    compress_string,
    reverse_words,
)
from pytekt.algorithms.searching.search import kmp_search

print("Longest palindrome:    ", longest_palindrome_substring("babad"))
print("Reverse words:         ", reverse_words("the sky is blue"))
print("Is 'listen' anagram of 'silent':", is_anagram("listen", "silent"))
print("String run-length info: ", compress_string("aabcccccaaa"))
print("KMP search occurrences: ", kmp_search("ABABDABACDABABCABAB", "ABABCABAB"))
```

## 2. Data Compression Routines


```python
from pytekt.algorithms.strings.compression import (
    run_length_encode,
    run_length_decode,
    delta_encode,
    delta_decode,
    base64_encode,
    base64_decode,
)

# Run Length Encoding (RLE)
raw_str = "AAAAABBBCCDAA"
encoded_rle = run_length_encode(raw_str)
print("RLE Encoded: ", encoded_rle)
print("RLE Decoded: ", run_length_decode(encoded_rle))

# Delta Encoding for numeric sequences
series = [100, 103, 105, 110, 115]
deltas = delta_encode(series)
print("Delta Encoded:", deltas)
print("Delta Decoded:", delta_decode(deltas))
```
