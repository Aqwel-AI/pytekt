# Graph Algorithms & Network Analysis

Demonstrates traversal, pathfinding, minimum spanning trees, and centrality from `pytekt.algorithms.graphs`.


## 1. Graph Traversal & Topological Ordering


```python
from pytekt.algorithms.graphs import (
    bfs,
    dfs,
    toposort,
    connected_components,
    shortest_path_unweighted,
)

# Directed Acyclic Graph (DAG)
dag = {
    "prep": ["cook", "clean"],
    "cook": ["eat"],
    "clean": ["eat"],
    "eat": ["sleep"],
    "sleep": []
}

print("BFS Traversal:        ", bfs(dag, "prep"))
print("DFS Traversal:        ", dfs(dag, "prep"))
print("Topological Sort:     ", toposort(dag))
print("Shortest Path prep->sleep:", shortest_path_unweighted(dag, "prep", "sleep"))
```

## 2. Weighted Shortest Paths (Dijkstra, Bellman-Ford, Floyd-Warshall)


```python
from pytekt.algorithms.graphs import dijkstra, bellman_ford

weighted_graph = {
    "A": [("B", 2.0), ("C", 5.0)],
    "B": [("C", 1.0), ("D", 4.0)],
    "C": [("D", 2.0)],
    "D": []
}

print("Dijkstra distances from A:", dijkstra(weighted_graph, "A"))
```

## 3. Spanning Trees & Centrality (MST & PageRank)


```python
from pytekt.algorithms.graphs import kruskal_mst, prim_mst, pagerank

weighted_edges_graph = {
    "A": [("B", 1.0), ("C", 3.0)],
    "B": [("A", 1.0), ("C", 2.0)],
    "C": [("A", 3.0), ("B", 2.0), ("D", 4.0)],
    "D": [("C", 4.0)],
}

print("Kruskal MST: ", kruskal_mst(weighted_edges_graph))
print("Prim MST:    ", prim_mst(weighted_edges_graph))

# PageRank on link network
web_graph = {
    "pageA": ["pageB", "pageC"],
    "pageB": ["pageC"],
    "pageC": ["pageA"],
    "pageD": ["pageC"]
}
print("PageRank:    ", pagerank(web_graph))
```
