# Dynamic Programming & Greedy Algorithms

Explore algorithmic design paradigms from `pytekt.algorithms.paradigms`.


## 1. Dynamic Programming Primitives

Solve classic optimization and string problems with optimal substructure and overlapping subproblems.


```python
from pytekt.algorithms.paradigms.dynamic_programming import (
    knapsack_01_max_value,
    lcs_length,
    edit_distance,
    coin_change_min_coins,
    lis_length,
    matrix_chain_order,
)

# 0/1 Knapsack
weights = [1, 2, 3]
values = [6, 10, 12]
capacity = 5
print("0/1 Knapsack Max Value:", knapsack_01_max_value(weights, values, capacity))

# Longest Common Subsequence (LCS)
print("LCS length ('abcde', 'ace'):", lcs_length("abcde", "ace"))

# Levenshtein Edit Distance
print("Edit distance ('kitten', 'sitting'):", edit_distance("kitten", "sitting"))

# Longest Increasing Subsequence
print("LIS length ([10, 9, 2, 5, 3, 7, 101, 18]):", lis_length([10, 9, 2, 5, 3, 7, 101, 18]))

# Coin Change Problem
print("Min coins for 11 with [1, 2, 5]:", coin_change_min_coins([1, 2, 5], 11))
```

## 2. Greedy Choice Strategies

Apply locally optimal choices that lead to globally optimal solutions.


```python
from pytekt.algorithms.paradigms.greedy import (
    fractional_knapsack,
    activity_selection,
    gas_station,
)

# Fractional Knapsack
w = [10, 20, 30]
v = [60, 100, 120]
print("Fractional Knapsack Value (capacity=50):", fractional_knapsack(w, v, 50))

# Activity Selection (interval scheduling)
intervals = [(1, 4), (3, 5), (0, 6), (5, 7), (8, 9), (5, 9)]
print("Max non-overlapping activities:", activity_selection(intervals))
```
