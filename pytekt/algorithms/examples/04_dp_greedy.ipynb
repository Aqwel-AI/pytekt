# Dynamic programming and greedy

Classic DP and greedy problems from the algorithm catalog.

```python
from aion.algorithms import get_algorithm

knapsack = get_algorithm("knapsack_01_max_value")
print("0/1 knapsack max value:", knapsack([1, 2, 3], [6, 10, 12], 5))

lcs = get_algorithm("lcs_length")
print("LCS length:", lcs("abcde", "ace"))

activity = get_algorithm("max_meetings")
print("activities:", activity([(1, 4), (3, 5), (0, 6), (5, 7), (8, 9), (5, 9)]))
```
