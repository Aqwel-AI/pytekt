# Mathematics, Numerical Methods & Computational Geometry

Explore mathematical primitives from `pytekt.algorithms.mathematics`.


## 1. Number Theory & Primes


```python
from pytekt.algorithms.mathematics.math_number_theory import (
    gcd,
    lcm,
    is_prime,
    sieve_primes,
    extended_gcd,
    catalan_number,
)

print("GCD(48, 18):          ", gcd(48, 18))
print("LCM(48, 18):          ", lcm(48, 18))
print("Is 97 prime?          ", is_prime(97))
print("Primes up to 30:      ", sieve_primes(30))
print("Catalan number C(5):  ", catalan_number(5))
```

## 2. Numerical Root Finding & Calculus


```python
from pytekt.algorithms.mathematics.numerical import (
    bisection_root,
    trapezoidal_integration,
    simpson_integration,
)

# Root finding for f(x) = x^2 - 2 (solving sqrt(2))
f = lambda x: x**2 - 2
root = bisection_root(f, 1.0, 2.0, tol=1e-6)
print("Bisection root (sqrt(2)):", round(root, 6))

# Numerical integration for f(x) = x^2 from 0 to 1 (exact = 1/3)
print("Simpson rule integral:  ", round(simpson_integration(lambda x: x**2, 0.0, 1.0, n=100), 6))
```

## 3. Computational Geometry


```python
from pytekt.algorithms.mathematics.geometry import (
    convex_hull,
    polygon_area,
    closest_pair_distance,
)

points = [(0, 0), (1, 1), (2, 2), (0, 2), (2, 0), (1, 0.5)]
hull = convex_hull(points)
print("Convex Hull vertices:  ", hull)
print("Polygon area:          ", polygon_area([(0, 0), (4, 0), (4, 3), (0, 3)]))
print("Closest pair distance: ", closest_pair_distance([(0, 0), (5, 5), (1, 1), (10, 10)]))
```
