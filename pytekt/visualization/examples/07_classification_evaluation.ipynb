# Classification & Model Evaluation Diagnostics

This notebook demonstrates evaluation and model diagnostics from `pytekt.visualization.ml_eval`.


## 1. Receiver Operating Characteristic (ROC) & Precision-Recall (PR) Curves

Evaluate discrimination capability and tradeoff between precision and recall on imbalanced datasets.


```python
from pytekt.visualization.ml_eval import plot_roc_curve, plot_pr_curve

y_true = [0, 0, 0, 0, 1, 0, 1, 1, 0, 1, 1, 1]
y_score = [0.05, 0.12, 0.28, 0.35, 0.42, 0.48, 0.65, 0.72, 0.78, 0.85, 0.91, 0.95]

fig_roc = plot_roc_curve(y_true, y_score, title="Classifier ROC Curve", show=False)
fig_pr = plot_pr_curve(y_true, y_score, title="Classifier Precision-Recall Curve", show=False)
```

## 2. Probability Calibration (Reliability Diagram)

Assess whether predicted probabilities match real-world observed frequencies.


```python
from pytekt.visualization.ml_eval import plot_calibration_curve

fig_cal = plot_calibration_curve(
    y_true=y_true,
    y_prob=y_score,
    n_bins=5,
    title="Model Reliability / Calibration Diagram",
    show=False,
)
```

## 3. Class Distribution & Residual Analysis

Inspect dataset balance and regression/scoring residuals.


```python
from pytekt.visualization.ml_eval import plot_class_distribution, plot_residuals

# Class Distribution
labels = ["Negative", "Neutral", "Positive"]
y_classes = [0, 1, 1, 2, 2, 2, 2, 0, 1, 2, 0, 1]
fig_dist = plot_class_distribution(y_classes, labels=labels, title="Sentiment Class Balance", show=False)

# Residual Analysis
y_actual = [10.5, 14.2, 18.0, 22.5, 27.1, 31.8]
y_predicted = [10.2, 14.8, 17.5, 23.0, 26.5, 32.4]
fig_res = plot_residuals(y_actual, y_predicted, title="Regression Residuals Analysis", show=False)
```
