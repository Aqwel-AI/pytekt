# Seaborn plots

Requires `pip install aqwel-pytekt[viz]`.

```python
import numpy as np
from pytekt.visualization import set_pytekt_style, sns_heatmap, sns_regplot

set_pytekt_style()
data = np.arange(16).reshape(4, 4)
sns_heatmap(data, title="Correlation-style heatmap", show=True)

x = np.linspace(0, 10, 30)
y = 2 * x + np.random.randn(30)
sns_regplot(x, y, title="Linear fit", show=True)
```
