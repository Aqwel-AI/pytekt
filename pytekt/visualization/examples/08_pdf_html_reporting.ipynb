# Report Generation: Multi-Page PDF & HTML Export

Demonstrates compiling figures into multi-page PDF documents and embedding figures in HTML reports using `pytekt.visualization.reporting`.


## 1. Create Sample Scientific Figures


```python
from pytekt.visualization.one_d import plot_array, plot_histogram
from pytekt.visualization.two_d import plot_matrix_heatmap
from pytekt.visualization.ml_eval import plot_training_history

# Figure 1: 1D Signal
fig1 = plot_array([10, 12, 15, 14, 18, 22, 20, 25], title="Signal Monitoring", show=False)

# Figure 2: Distribution
fig2 = plot_histogram([1, 2, 2, 3, 3, 3, 4, 4, 5, 5, 5, 5, 6], bins=6, title="Value Distribution", show=False)

# Figure 3: Matrix Heatmap
fig3 = plot_matrix_heatmap([[1, 2, 3], [4, 5, 6], [7, 8, 9]], title="Correlation Matrix", show=False)

# Figure 4: Training Loss
fig4 = plot_training_history({"loss": [0.8, 0.5, 0.3], "val_loss": [0.9, 0.6, 0.4]}, show=False)

figures = [fig1, fig2, fig3, fig4]
```

## 2. Compile Figures into a Multi-Page PDF Report


```python
import tempfile
from pathlib import Path
from pytekt.visualization.reporting import save_figures_pdf

with tempfile.TemporaryDirectory() as tmpdir:
    pdf_path = Path(tmpdir) / "experiment_report.pdf"
    save_figures_pdf(figures, str(pdf_path))
    print(f"Generated multi-page PDF at: {pdf_path} (size: {pdf_path.stat().st_size} bytes)")
```

## 3. Convert Figures to Base64 HTML `<img>` Tags for Web Dashboards


```python
from pytekt.visualization.reporting import figures_to_html_img_tags

html_tags = figures_to_html_img_tags(figures, fmt="png")
print(f"Generated {len(html_tags)} base64 HTML img tags.")
print("Sample tag prefix:", html_tags[0][:60], "...")
```
