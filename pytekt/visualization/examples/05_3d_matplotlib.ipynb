# Matplotlib 3D plots

Advanced static 3D visualizations from `aion.visualization.three_d`.

```python
import numpy as np
from aion.visualization import plot_3d_surface, plot_3d_wireframe, plot_3d_trajectory

u = np.linspace(-2, 2, 30)
v = np.linspace(-2, 2, 30)
X, Y = np.meshgrid(u, v)
Z = np.sin(np.sqrt(X**2 + Y**2))
plot_3d_surface(u, v, Z, title="Sinc surface", show=True)
plot_3d_wireframe(u, v, Z, title="Wireframe", show=True)

t = np.linspace(0, 4 * np.pi, 200)
plot_3d_trajectory(np.cos(t), np.sin(t), t, title="Helix", show=True)
```
