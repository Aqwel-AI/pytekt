# Plotly interactive 3D

Requires `pip install aqwel-pytekt[viz3d]`.

```python
import numpy as np
from pytekt.visualization import plotly_3d_scatter, plotly_3d_surface, show_plotly

fig = plotly_3d_scatter([1, 2, 3], [4, 5, 6], [7, 8, 9], title="3D scatter")
show_plotly(fig)

x = np.linspace(0, 1, 20)
y = np.linspace(0, 1, 20)
X, Y = np.meshgrid(x, y)
fig2 = plotly_3d_surface(x, y, X * Y, title="Interactive surface")
show_plotly(fig2)
```
